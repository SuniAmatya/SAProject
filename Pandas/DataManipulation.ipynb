apply()- Applies a function to a certain axis. Default axis = 0 (columnwise)

```python
def replace_with_gender(row):
    if row.Sex == "male":
        value = "Mr"
    else:
        value = "Mrs"

    return row.initials.replace("Dr", value, True)
```

```python
titanic_df.initials = titanic_df.apply(replace_with_gender, axis = 1)
```

```python
titanic_df.initials.value_counts()
```

```python
group_by_initials = titanic_df.groupby(by ="initials")
```

```python
titanic_df.groupby(["initials"]).Age.describe()
```

```python
med_age= group_by_initials["Age"].transform("median")
# calculates median age for each unique group and broadcasts those median values back to match size of your original DataFrame.
# here it reflects median of each initial and broadcasts the median to match the initial of the row
# Mr median is 30, so all Mr have med_age 30. 
# Mrs median is 35, so all Mr have med_age 35. 
```

```python
med_age
```

```python
titanic_df_fillAge_initials= titanic_df.copy()

titanic_df_fillAge_initials.loc[:, "Age"] = titanic_df_fillAge_initials.Age.fillna(
    value=med_age
)
```

Outlier Handling

1. IQR - Inter Quartile Range

```python
titanic_df_fillAge_initials.Age.plot(kind="box", vert= False)
```

```python
Q1 = titanic_df_fillAge_initials.Age.quantile(q=0.25)
Q3 = titanic_df_fillAge_initials.Age.quantile(q=0.75)

IQR = Q3 - Q1

maximum = Q3 + (1.5 *IQR)
minimum = Q1 - (1.5 * IQR)
```

```python
maximum
```

```python
minimum
```

```python
titanic_df_fillAge_outlierAge= titanic_df_fillAge_initials[
    (titanic_df_fillAge_initials.Age > maximum) | (titanic_df_fillAge_initials.Age < minimum)
]
```

```python
titanic_df_fillAge_outlierAge.head()
```

```python
titanic_df.Fare.plot(kind="box", vert= False, figsize=(12,2))
```

```python
Q1 = titanic_df.Fare.quantile(q=0.25)
Q3 = titanic_df.Fare.quantile(q=0.75)

IQR = Q3 - Q1

maximum = Q3 + (3 *IQR)
minimum = Q1 - (0.25 * IQR)
```

```python
minimum
```

```python
titanic_df_outlierFare= titanic_df[
    (titanic_df.Fare > maximum) | (titanic_df.Fare < minimum)
]
```
