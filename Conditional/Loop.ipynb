```python
For Loop
Syntax:
    for variable in sequence:
        pass
```

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"{fruit} - Rs. 100")
```

```python
for i in range(10):
    print(i)
```

```python
star="*"
for i in range(1,6):
    print (star * i)
```

```python
#Nested for loop
adj =["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    for ad in adj:
        print(ad, fruit)
```

```python
number= 4.315
for i in range(1,11):
    print(f"{number} x {i} = {number*i}")
```

```python
#Multiple arguments in loop
Can run only if they are of same size
```

```python
person={
    "name" :"Robin",
    "age" : 14
}
for item in person.items():
    print(item)
```

```python
coordinate = [(1,5,9), (25,45,7), (10, 11, 14)]
for x, y, z in coordinate:
    print(f"{x}, {y} , {z}")
```
