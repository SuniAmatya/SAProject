Class: A blueprint or template for creating objects. It defines structural guidelines and shared behavior.
Object: A specific instance generated from a class containing real, localized data.
Variable: Attribute
Function: Properties
__init__: The constructor method that executes automatically when a new object is initialized.
self: A reference pointer targeting the specific object instance currently executing the method

Syntax:
    class ClassName:
        def methods:
        ...

```python
class Calculator:
    def __init__(self, num1, num2):
        print("Making Objects...")
        self.num1= num1
        self.num2= num2

    def add(self):
        return self.num1 + self.num2
    
    def multiply(self, num3: int) -> int| float:
        return self.num1 * self.num2 * num3
```

```python
calc = Calculator(10,6)
print(calc.multiply(6))
calc.add()
```
