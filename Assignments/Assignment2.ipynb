```python
# 1. Write a Program that Uses Functions write_to_file and read_from_file:
# write_to_file(filename, content): Writes content to a file named filename. If the file doesn't exist, it should be created.
# read_from_file(filename): Reads and prints the content of a file named filename. 
# Call write_to_file to write "Hello, Python!" to a file named "greetings.txt", then call read_from_file to read and print the content of this file.
```

```python
def write_to_file(filename, content):
    with open(filename, mode='w') as file:
        file.write(content)
        print(f"Successfully wrote to {filename}")

def read_from_file(filename):
    with open(filename, mode= "r") as file:
        print(file.read())

write_to_file("greeting.txt", "Hello, Python!")
read_from_file("greeting.txt")
```

```python
#Control Flow with Nested Loops and Complex Logic
# Write a Python program that simulates a number guessing game:
play_again= True
import random
while play_again:
    n = 0
    secret_num= random.randint(1,100)
    guess= 0
    while guess != secret_num and n < 7:
        n += 1
        guess= int(input("Enter a number (1-100):"))
        if guess > secret_num:
            print(f"{guess} is too high.")
        elif guess < secret_num:
            print(f"{guess} is too low")
        else:
            print(f"You got it! The number is {secret_num}")
    
    if guess!= secret_num:
        print(f"{secret_num} is the number. Do you want to try again?")

    again= input("Do you want to play again (y/n)").lower()
    if again != "y":
        play_again = False
        print("Thank you for playing")
```

```python
#Write a Python program to check whether a given number is an Armstrong number or not.
num= int(input("Enter a number"))
if num >0:
    string_num= str(num)
    length= len(string_num)
    armstrong = 0
    for digit in string_num:
        armstrong += int(digit) ** length
    if armstrong == num:
        print(f"Yes, it's an Armstrong number.{num}")
    else:
        print(f"No, it's not an Armstrong number.{num}")
else:
    print("Number should be a positive integer.")
```

```python
num= int(input("Enter a number"))
copied= num
if num >0:
    string_num= str(num)
    length= len(string_num)
    armstrong = 0
    
    while num > 0:
        digits = num % 10
        armstrong += digits ** length
        num= num // 10
        
    if armstrong == copied:
        print(f"Yes, it's an Armstrong number.{copied}")
    else:
        print(f"No, it's not an Armstrong number.{copied}")
else:
    print("Number should be a positive integer.")
```

```python
#Write a Python program that iterates through integers from 1 to 50. 
# For each multiple of three, print "Fizz" instead of the number; for each multiple of five, print "Buzz". 
# For numbers that are multiples of both three and five, print "FizzBuzz".


for i in range(1,50):
    if i % 3 ==0 and i % 5 ==0:
        print("FizzBuzz")
    elif i % 3 ==0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)    
```
