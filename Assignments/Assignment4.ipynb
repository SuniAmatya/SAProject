Math Tutor Using Random, Arithmetic Operators, and OOP
Scenario: Create a MathTutor class that generates random math questions using random and math operators.

Track correct and incorrect answers. 

Provide a score at the end.

Handle invalid input using exception handling.

Concepts: Random, Math, Exception Handling, Classes

```python
import random

class MathTutor():
    def __init__(self):
        self.correct = 0
        self.incorrect = 0
        self.operators = ["+", "-", "*", "**", "/", "//", "%", "sqrt"]

    def Questions(self):
        num1 = random.randint(1, 100)
        num2 = random.randint(1, 100)
        operator = random.choice(self.operators)

        question = f"{num1} {operator} {num2}"
        answer = eval(question)

        return question, answer

    def start_MathTutor(self):
        print("Welcome to MathTutor")

        for i in range(5):
            question, answer = self.Questions()

            print(f"\nQuestion {i + 1}: {question}")

            try:
                input_answer= float(input("Your Answer: "))

                if input_answer == answer:
                    print("You got it!")
                    self.correct += 1
                else:
                    print(f"Aww! You almost made it. The answer was {answer}")
                    self.incorrect += 1

            except ValueError:
                print("Invalid input. Enter numbers only!")
                self.incorrect += 1

        print(f"\n{"*" * 5}The End{"*" * 5}")
        print("Correct:", self.correct)
        print("Incorrect:", self.incorrect)

```

```python
play = MathTutor()
play.start_MathTutor()
```

```python
play.start_MathTutor()
```

Exception Handling Scenario: Online Age-Restricted Service

Scenario: You’re building a sign-up system for an online movie rental platform. Some movies are age-restricted (18+). You need to ensure proper validation and error handling during user registration.

Task:

Create a custom exception class called UnderageError that inherits from Exception.

Write a function register_user() that:

Takes a user’s name and age as input.

Raises UnderageError if the user is under 18.

Otherwise, prints a welcome message.

Wrap the function call in a try block and handle the exception.

Use else to confirm successful registration and finally to always print “Thank you for using MovieTime!” regardless of outcome.

Also try to validate if the age input is numeric. Raise a ValueError if not, and handle it separately.

```python
class UnderageError(Exception):
    pass

def register_user(user_name:str, age: int):
    if not isinstance(age, int):
        raise ValueError("Age must be a whole number")

    if age < 18:
        raise UnderageError("Registration Failed. User must be older than 18 to watch movie.")

    print(f"Welcome to MovieTime: Rent-a-Movie, {user_name}")

try:
    register_user(
        input("Enter your name:"), 
        int(input("Enter your age:"))
    )
except UnderageError as e:
    print("Registration Failed!")
except ValueError as e:
    print(f"Invalid Input {e}")
else:
    print("Thank you for using MovieTime!")
```

We want to build an online shopping cart system that allows users to add products to their cart, calculate the total cost, apply discounts, and generate an invoice. The system should include the following functionalities:

Adding products to the cart

Removing products from the cart

Calculating the total cost

Applying discounts based on user type

Generating an invoice

1. Create the Product class
We create a basic Product class with attributes for the product name and price.

```python
class Product:
    def __init__(self, product_name, price: int):
        self.product_name = product_name
        self.price = price
```

2. Implement the User class:
In this step, we create a User class with attributes for the user's name and whether they are a premium member.

```python
class User:
    def __init__(self, user_name, membership = False):
        self.user_name = user_name
        self.membership = membership
```

3. Create the ShoppingCart class:
In this step, we create a ShoppingCart class with methods for adding and removing products from the cart, as well as calculating the total cost of the items in the cart.

Note: Define calculate_total_cost method in the ShoppingCart class, that applies a 10% discount to the total cost if you are premium User.

```python
class ShoppingCart:
    def __init__(self, user):
        self.user = user
        self.cart_products= []

    def add_cart(self, product):
        self.cart_products.append(product)
        print(f"{product.product_name} has successfully been added to cart.")

    def remove_product(self, product):
        if product in self.cart_products:
            self.cart_products.remove(product)
            print(f"{product.product_name} has successfully been removed from cart.")
        else:
            print(f"{product.product_name} is not in the cart.")

    def total_cost(self):
        total = sum(product.price for product in self.cart_products)

        if self.user.membership:
            total *= 0.90

        return total

    def generate_invoice(self):
        print(f"{"=" * 7} TAX INVOICE {"=" * 7}")
        print("Name: ", self.user.user_name)
        print(f"{"_" * 30}")

        for product in self.cart_products:
            print(f"{product.product_name} {" " * 5} {product.price}")

        gross_total = sum(product.price for product in self.cart_products)

        if self.user.membership:
            discount=  gross_total * 0.10
        else:
            discount = 0

        print(f"{"_" * 30}")
        print(f"Gross Total {" " * 5} {gross_total}")
        print(f"Discount {" " * 10} {discount}")
        print(f"Net Total {" " * 8} {gross_total - discount}")
```

4. Testing the functionality
Now that we have implemented the necessary classes and methods, let's test our online shopping cart system:

```python
item1 = Product("Laptop", 90000)
item2 = Product("TV", 180000)
item3 = Product("Projector", 20000)

user1= User("Rae", True)
user2= User("Joe", False)
```

```python
cart1= ShoppingCart(user1)
cart1.add_cart(item1)
cart1.add_cart(item2)
cart1.add_cart(item3)

total1= cart1.total_cost()

print("Your total: ", total1)
```
